# VARIABLES Y TIPOS DE DATOS

### Números

Los números son de tres tipos:
- int
- float
- complex

#### int

Números positivos o negativos sin decimales

#### float

Números positivos o negativos que contienen uno o más decimales

#### complex

Números complejos tiene una parte real y otra imaginaria. Para indicar la parte imaginaria se utiliza una j.

```python
x = 4+3j
y = 3j
z = -5j
```

### Conversión de tipos

Se pueden convertir el tipo utilizando los métodos int(), float(), complex()

```python
x = 1    # int
y = 2.3  # float
z = 4j   # complex

#convertir de int a float:
a = float(x)

#convertir de float a int:
b = int(y)

#convertir de int a complex:
c = complex(x)

print(x)
print(y)
print(z)

print(a)
print(b)
print(c)

print(type(a))
print(type(b))
print(type(c))
```
