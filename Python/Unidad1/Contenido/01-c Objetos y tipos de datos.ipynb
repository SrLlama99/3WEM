### Objetos y tipos de datos en Python
###### (realizado IA)
### 1. Todo es un objeto

En Python, prácticamente todo lo que utilizamos durante la ejecución de un programa es un objeto.

Por ejemplo:

edad = 25
precio = 19.95
nombre = "Laura"
dato = None


Los valores:

25
19.95
"Laura"
None


son objetos.

Cada objeto tiene, entre otras características:

Un tipo.

Un valor.

Una identidad.

Podemos consultar su tipo mediante type():

edad = 25

print(type(edad))


Resultado:

<class 'int'>

### 2. ¿Qué es una variable?

En Python, una variable puede entenderse como un nombre que referencia un objeto.

Por ejemplo:

edad = 25


Podemos imaginarlo así:

edad ───────► 25

                 │

                 └── objeto de tipo int


El nombre edad hace referencia al objeto 25.

No debemos pensar necesariamente que la variable es una "caja" que contiene el número.

Una forma más precisa de expresarlo es:

Una variable es un nombre asociado a un objeto.

### 3. Las variables pueden cambiar de tipo

Python utiliza tipado dinámico.

Esto significa que una misma variable puede hacer referencia a objetos de diferentes tipos durante la ejecución:

dato = 100
dato = 100.5
dato = "100.5"
dato = None


En cada momento:

dato ──► 100       int


después:

dato ──► 100.5     float


después:

dato ──► "100.5"   str


y finalmente:

dato ──► None      NoneType


El nombre dato sigue siendo el mismo, pero el objeto al que hace referencia cambia.

### 4. Los tipos de datos también son objetos

En Python, los tipos también son objetos.

Por ejemplo:

x = int


Aquí x hace referencia al objeto que representa el tipo int.

Podemos comprobarlo:

print(type(int))


El resultado es:

<class 'type'>


Lo mismo ocurre con:

print(type(str))
print(type(float))
print(type(None))


Aunque None tiene una particularidad: None es un objeto cuyo tipo es NoneType.

### 5. Objetos int

Los números enteros pertenecen al tipo int.

edad = 25
cantidad = -10
puntos = 0


Podemos consultar su tipo:

print(type(edad))


Resultado:

<class 'int'>


Los objetos int permiten realizar operaciones matemáticas:

a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a // b)
print(a % b)

### 6. Objetos float

Los números con parte decimal pertenecen normalmente al tipo float.

precio = 19.95
altura = 1.82
temperatura = -3.5


Por ejemplo:

print(type(precio))


Resultado:

<class 'float'>


Una operación entre un int y un float normalmente produce un float:

x = 10
y = 2.5

resultado = x + y

print(resultado)
print(type(resultado))


Resultado:

12.5
<class 'float'>

### 7. Objetos str

Las cadenas de texto pertenecen al tipo str.

nombre = "Carlos"
ciudad = "Madrid"
codigo = "28001"


Es importante entender que:

codigo = "28001"


es texto, aunque esté formado exclusivamente por números.

Por tanto:

type(codigo)


produce:

<class 'str'>


Esto es diferente de:

codigo = 28001


que es un int.

### 8. Objetos None

None representa normalmente la ausencia de un valor.

resultado = None


Su tipo es:

type(resultado)


Resultado:

<class 'NoneType'>


None no es lo mismo que:

0


ni que:

""


ni que:

"None"


Son objetos diferentes:

a = None
b = 0
c = ""
d = "None"


Sus tipos son:

a → NoneType
b → int
c → str
d → str

### 9. None no es un texto

Esta diferencia es especialmente importante:

a = None
b = "None"


En el primer caso:

a → ausencia de valor


En el segundo:

b → texto formado por los caracteres N-o-n-e


Por eso:

a == b


produce:

False

### 10. Identidad de los objetos

Los objetos también tienen una identidad.

Podemos consultar la identidad mediante id():

x = 100

print(id(x))


id() devuelve un identificador asociado al objeto durante su vida.

Podemos utilizar is para comprobar si dos variables hacen referencia al mismo objeto:

a = None
b = None

print(a is b)


Con None, la comprobación habitual es:

if dato is None:
    print("No hay datos")


Es preferible utilizar:

is None


en lugar de:

== None


para comprobar específicamente si una variable referencia a None.

### 11. Valor, tipo e identidad

Podemos resumir un objeto mediante tres conceptos:

OBJETO

 ├── valor

 ├── tipo

 └── identidad


Por ejemplo:

x = 25


Podemos consultar:

print(x)          # valor

print(type(x))    # tipo

print(id(x))      # identidad


Conceptualmente:

x ─────────► objeto

             │

             ├── valor: 25

             ├── tipo: int

             └── identidad: ...

### 12. Asignar una nueva referencia

Observa:

x = 10
y = x


Podemos imaginar:

x ───┐

     ├──► 10

y ───┘


Los dos nombres hacen referencia al objeto correspondiente.

Después:

x = 20


Ahora:

x ───► 20

y ───► 10


La asignación de x no modifica automáticamente y.

### 13. Cambiar la referencia no es modificar necesariamente el objeto

Consideremos:

x = 10

x = 20


No estamos "transformando" el objeto 10 en 20.

Estamos haciendo dos asignaciones:

Primero:

x ───► 10


Después:

x ───► 20


El nombre x deja de hacer referencia al objeto anterior y pasa a referenciar otro objeto.

Esta distinción será especialmente importante cuando estudiemos objetos mutables e inmutables.

### 14. Conversión de objetos

Python proporciona funciones para crear objetos de determinados tipos a partir de otros valores.

Por ejemplo:

x = int("25")

y = float("3.14")

z = str(100)


Tenemos:

"25"  ──► int()   ──► 25

"3.14" ──► float() ──► 3.14

100   ──► str()   ──► "100"


Podemos comprobar los tipos:

print(type(x))

print(type(y))

print(type(z))

### 15. No todas las conversiones son posibles

Por ejemplo:

int("25")


funciona.

Pero:

int("25.5")


produce un error porque "25.5" no representa directamente un literal entero válido para int().

Sin embargo:

float("25.5")


sí funciona.

Esto muestra que las conversiones dependen del tipo de objeto y de su contenido.

## 16. Idea fundamental

Para comprender Python a un nivel más profundo, conviene pensar en:

NOMBRE ─────────► OBJETO

                    │

                    ├── tipo

                    ├── valor

                    └── identidad


Por ejemplo:

nombre = "Ana"

edad = 25

altura = 1.68

observaciones = None


Conceptualmente:

nombre ─────────► "Ana"       → str

edad ───────────► 25          → int

altura ─────────► 1.68        → float

observaciones ──► None        → NoneType


Y, gracias al tipado dinámico, podemos posteriormente hacer:

edad = "veinticinco"


Ahora:

edad ───────────► "veinticinco" → str


El nombre edad no tiene un tipo fijo. El objeto al que hace referencia sí tiene un tipo.

### Conceptos que debes recordar

| Concepto | Significado |
|---|---|
| **Objeto** | Entidad que existe durante la ejecución del programa y que tiene un tipo, un valor y una identidad. |
| **Variable** | Nombre que referencia a un objeto. |
| **`int`** | Tipo de dato utilizado para representar números enteros. Ejemplo: `25` |
| **`float`** | Tipo de dato utilizado para representar números decimales. Ejemplo: `3.14` |
| **`str`** | Tipo de dato utilizado para representar cadenas de texto. Ejemplo: `"Hola"` |
| **`None`** | Valor especial que representa normalmente la ausencia de un valor. |
| **`NoneType`** | Tipo de dato del objeto `None`. |
| **`type()`** | Función que permite conocer el tipo de un objeto. |
| **`id()`** | Función que permite consultar la identidad de un objeto. |
| **`is`** | Operador que comprueba si dos variables hacen referencia al mismo objeto. |
| **`==`** | Operador que compara el valor de dos objetos. |
| **Tipado dinámico** | Una variable puede hacer referencia a objetos de diferentes tipos durante la ejecución. |
| **Conversión de tipos** | Transformación de un objeto a otro tipo mediante funciones como `int()`, `float()` o `str()`. |
| **Referencia** | Relación entre un nombre y el objeto al que hace referencia. |





No pienses únicamente:

"La variable x es un entero."

Es más preciso pensar:

"x es un nombre que actualmente referencia a un objeto de tipo int."

Por ejemplo:

x = 10

x = "hola"

x = None


El nombre x permanece, pero el objeto al que hace referencia cambia.

Este modelo mental será fundamental para entender posteriormente mutabilidad, referencias, listas, diccionarios, funciones, argumentos y gestión de memoria en Python.
