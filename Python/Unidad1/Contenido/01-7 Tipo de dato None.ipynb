## Tipo de dato NoneType

"None" es un tipo de datos especial en Python que representa la ausencia de valor. El tipo de datos es "NoneType". None representa un objeto único que indica la ausencia de valor.

```python
x = None
type(x) # NoneType
```

### Uso
Se utiliza para:
- Crear una variable sin valor. 
- valores por defecto en funciones
- Retorno en funciones que no devuelven nada explícitamente. 

```python
resultado = None # Inicializar variables
#Valores por defecto en funciones
def saludo(nombre=None):
    if nombre == None:
        print("hola")
    else:
        print("hola",nombre)

saludo()
saludo("Pepe")
        

#Retorno de funciones que no devuelve nada expícitamente
def sumar(a,b):
    c = a + b
    
print(sumar(1,2))
```

### Comparación
Para la comparación se utiliza el operador "is" o el "is not". "is" verifica si dos variables se refieren al mismo objeto en memoria.

```python
x = None
print( x is None)
print( x is not None)
```

```python
a = None
b = None
id(a)
id(b)
```

"==" es el operador verifica la equivalencia de valor. Aunque None == None es True, usar "==" con None es menos eficiente que "is". 

### Diferencia con otros "valores vacíos"
Es crucial entender que "None" es diferente de otros valores que se pueden considerar "vacíos": 
- "None": Indica la ausencia de un valor. 
- Cadena vacía (""): Es un objeto de tipo cadena con cero caracteres. 
- Lista vacía ([]): Es un objeto de tipo lista que no contiene elementos. 
- Cero (0) o False: Son valores específicos dentro de los tipos numéricos y booleanos, respectivamente.
