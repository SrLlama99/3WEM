Las variables son contenedores para almacenar un valor determinado. 
A las variables se les asigna valor con el símbolo =

Python no tiene comando para declarar variable. La variable se crea en el momento que se le asigna un valor por primera vez

```python
num1 = 1
```

```python
print(str1)
```

### Tipos de datos

Las variables almacenan
datos de diferentes tipos. El tipo permite hacer diferentes cosas.

- Text Type:	str
- Numeric Types:	int, float, complex
- Boolean Type:	bool
- Sequence Types:	list, tuple, range
- Mapping Type:	dict
- Set Types:	set, frozenset
- Binary Types:	bytes, bytearray, memoryview
- None Type:	NoneType
