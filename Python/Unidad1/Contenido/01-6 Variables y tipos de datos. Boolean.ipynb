### Booleans

Los booleanos permiten respresentar la certeza de una expresión. Tienen dos valores:
- True
- False

Cuando se comparan dos valores se evalúa la expresión y Python devuelve un booleano

Se utilizan para saber si una expresión es verdadera o falsa

```python
print(9<3)
```
