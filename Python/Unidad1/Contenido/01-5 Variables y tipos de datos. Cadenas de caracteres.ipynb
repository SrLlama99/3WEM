# VARIABLES Y TIPOS DE DATOS

### Cadenas de caracteres

Las cadenas de caracteres (string) en python están rodeadas por comillas dobles (") o simples ('). Permitir las dos formas facilta el poder poner " o ' como parte de la cadena de caracteres. 

La asignación de una cadena a una variable se realiza con el nombre de la variable seguido de un signo igual y la cadena

```python
x = "Hola"
y = 'Adios'
z = "Hola 'Pepe'"
print(x)
print(y)
print(z)
```

Se puede asignar una cadena multilínea a una variable usando tres comillas:

```python
a = """ Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed hacer eiusmod tempor incididunt
ut labore et dolore magna aliqua."""
print(a)
```

```python
a = ''' Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed hacer eiusmod tempor incididunt
ut labore et dolore magna aliqua. '''
print(a)
```

#### Indexación

Las cadenas son secuencias ordenadas de caracteres y pueden ser indexados. Se accede con el nombre de la cadena y el número entre corchetes.
El primer carácter tiene el índice 0, el siguiente el 1 y así sucesivamente. 

```python
var = "Hola"
```

```python
var[0]
```

```python
var[3]
```

```python
var[6]
```

El último carácter tendrá como índice la longitud de la cadena -1.
El índice puede tomar valores negativos siendo -1 el último carácter, -2 el penúltimo, etc. 

```python
var[-1]
```

```python
var[-3]
```

```python
var[-5]
```

#### Slicing

Se pueden extraer subcadenas utilizando sintaxis específica de indexación

s = var[n:m]

s es una subcadena de la variable var formada por los caracteres desde el ínide n al m (sin ser incluido el útlimo carácter). 

```python
s = var[0:1]
print(s)
```

```python
nombre = "Pepe García"
print(nombre[0:4])
```

```python
nombre[-6:-1]
```

Si no indicamos número, lee hasta el final

```python
nombre[-6:]
```

```python
nombre[5:]
```

```python
nombre[:4]
```

Uso de un índice fuera de rango generará un error

```python
nombre[50]
```

En el slicing un índice fuera de rango se gestiona satisfactoriamente

```python
nombre[0:44]
```

```python
nombre[33:]
```

```python
nombre[:40]
```

#### Stride

Se trata de una variante del slicing. Añade un tercer índice para indicar cuántos caracteres saltar hasta obtener el siguiente carácter.

```python
nombre[0:4:2]
```

#### Modificación Strings

No se puede modificar un parte del string asociada a una variable, se considera un tipo se considera inmutable. 

```python
nombre[3] = "a"
```

Si se necesita modificar una cadena de caracteres hay que crear una nueva

```python
nombre2 = nombre[0:3] + 'a' + nombre[4:]
nombre2
```

#### Strings múltiples líneas

Utilizando el carácter \n

```python
nombre = "Pepe\nGarcía\nMuñoz"
```

```python
print(nombre)
```

Utilizando  """

```python
nombre2 = """Pepe
García
Muñoz"""
```

```python
print(nombre2)
```
