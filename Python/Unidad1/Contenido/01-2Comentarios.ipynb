## COMENTARIOS

los comentarios se pueden utilizar:
   - para explicar el código Python.
   - para hacer que el código sea más legible.
   - para evitar la ejecución al probar el código.

Los comentarios comienzan por #

Los comentarios se pueden colocar al final de una línea y Python ignorará el resto de linia

# COMENTARIOS MULTILINEA

Python no tiene una sintaxis para comentarios multilínea

Dado que Python ignorará las cadenas literales que no están asignadas a una variable, se pueden utilizar las comillas triples y colocar el comentario dentro de ellas:

```python
"""
Este es un comentario
con más de una línea
para probar el funcionamiento.
"""
print("¡Hola Mundo!")
```
