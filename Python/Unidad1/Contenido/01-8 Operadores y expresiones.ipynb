### Operadores Lögicos

and    Devuelve True si todas las sentencias son True

or     Devuelve False si todas las sentencias son False

not    Devuelve False si el resultado es True y True si el resultado es False

```python
var1 = True
var2 = False
print( var1 and var2)
```

```python
print(var2 or var1)
```

```python
print( not var1)
```

### Operadores de pertenencia

in Devuelve True si una secuencia del valor especificado está presente en el objeto

```python
x = ["rojo", "azul"]

print("rojo" in x)
print("verde" in x)
```

not in Devuelve True si una secuencia del valor especificado no está presente en el objeto

```python
print("rojo" not in x)
print("verde" not in x)
```
