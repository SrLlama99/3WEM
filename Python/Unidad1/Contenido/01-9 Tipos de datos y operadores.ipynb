## Enteros. Asignación de variables
Los "int" son inmutables, no se pueden modificar en el lugar por lo que al ejecutar a += 1 se crea un nuevo objeto. El tipo de dato str y tuple también crean un nuevo objeto.

```python
a = 5
b = a
print(id(a)) # id de la variable a
a += 1
print(id(a)) # id de la variable a
print(id(b)) # id de la variable b
print(a)
print(b)
```

## Listas. Asignación de variables
Los "list" son mutables, por lo que al añadir elementos no se crea un nuevo objeto. El tipo de dato dict tampoco crea un nuevo objeto.

```python
a = [5]
b = a
print(id(a)) # id de la variable a
a.append(4)
print(id(a)) # id de la variable a
print(id(b)) # id de la variable b
print(a)
print(b)
```
