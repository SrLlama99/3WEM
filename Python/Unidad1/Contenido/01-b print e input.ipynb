## Funciones input() y print() en Python

En Python, input() y print() son dos funciones básicas que permiten interactuar con el usuario.

#### 1. La función print()

La función print() sirve para mostrar información en pantalla.

Por ejemplo:

```python
print("Hola, mundo")
```

También podemos mostrar el contenido de una variable:

```python
nombre = "Ana"
print(nombre)
```

Podemos mostrar varios valores utilizando comas:

```python
nombre = "Ana"
edad = 20

print("Nombre:", nombre)
print("Edad:", edad)
```

#### 2. La función input()
La función input() sirve para pedir información al usuario.

Por ejemplo:



```python
nombre = input("¿Cómo te llamas? ")
print("Hola", nombre)
```

#### 3. Importante: input() devuelve texto
Aunque el usuario escriba un número, input() lo recibe como texto.

Por ejemplo:

```python
edad = input("¿Cuántos años tienes? ")
```

Si escribimos 20, Python guarda "20" como texto, no como número.

Para convertirlo en un número entero utilizamos int():

```python
edad = int(input("¿Cuántos años tienes? "))
```

Ahora podemos realizar operaciones matemáticas:

```python
edad = int(input("¿Cuántos años tienes? "))

print("El año que viene tendrás", edad + 1, "años")
```

#### 4. Convertir datos con int() y float()
int()
Convierte un valor a un número entero:


```python
edad = int(input("Introduce tu edad: "))
```

#### 5. Ejemplo completo



```python
nombre = input("¿Cómo te llamas? ")

edad = int(input("¿Cuántos años tienes? "))

print("Hola", nombre)

print("Tienes", edad, "años")

print("El año que viene tendrás", edad + 1, "años")
```
