### for

Se utiliza la sentencia for para ejecutar sentencias repetidas un número de veces finito utiizando una secuencia de objetos como listas, tuplas, string y range.

```python
for <variable> in <iterable>:
    <sentencia(s)>

<iterable> es una colección de elementos
<variable> es una variable que recibirá uno de los elementos del objeto <iterable> en cada una de las iteraciones
<sentencia(s)> es el bloque de sentencias Python que se ejecutarán repetidamente
```

```python
colores = "verde, blanco"

for color in colores:
    print(color)
```

<iterable> Son objetos que pueden utilizarse para iteraciones. Los tipos de datos string, lista, tupla, diccionario, set o forzenset son objetos iterables.

#### else
En el bucle for se puede utilizar la cláusula else que se ejecutará cuando el iterador genere una excepción. 

```python
for color in colores:
    print(color)
else:
    print("Fin del for")
```

### Función range()
Define un rango de valores. Dado que Python funciona con indexación basada en cero, la secuencia comenzará en cero y terminará en el número especificado, es decir, n-1, donde n es el número especificado en la función range().
La función range() se puede representar de tres maneras diferentes:

```python
range(4) # El valor inicial es el 0
```

```python
range(0,4) # Se indica el valor inicial y el final
```

```python
range(0,4,1) # Se indica el valor inicial, el final y el incremento hasta alcanzar el valor final 
```

```python
type(range(4)) # tipo
```

#### Uso range en el for

```python
for i in range(4):
    print(i)
```

```python
for i in range(0,4):
    print(i)
```

```python
for i in range(0,4,1):
    print(i)
```

```python
for i in range(0,4,2):
    print(i)
```

### Break

Se utiliza para termianr de manera inmediata la ejecución de una estructura de control de flujo con for o while. Se pasará a ejecutar la siguiente línea de código que esté fura del for o while.

```python
colores = "verde, blanco"
for color in colores:
    break
    print(color) # no se llega a ejecutar
print("Fin for")
```

```python
var = True
it = 1 
while var:
    print("Es la iteración: ", it)
    it += 1
    break
    print("Después del break")
print("Fin while")
```

Al utilizar el break las sentencias en el else no se ejecutan

```python
colores = "verde, blanco"
for color in colores:
    break
    print(color) # no se llega a ejecutar
else:
    print("En el else") # No se ejecutan al terminar con un break
print("Fin for")
```

```python
var = True
it = 1 
while var:
    print("Es la iteración: ", it)
    it += 1
    break
    print("Después del break")
else:
    print("En el else")
print("Fin while")
```

### continue
Con esta sentencia se termina la iteración actual. 

```python
colores = "verde, blanco"
for color in colores:
    if color == "b":
        continue
    print(color) # se ejecuta cuando el color es distinto de verde
print("Fin for")
```

```python
var = True
it = 1 
while var:
    it += 1
    if it < 5:
        continue
        print("Es la iteración: ", it) # no se ejecuta
    else:
        print("Es la iteración: ", it)
        var = False
print("Fin while")
```
