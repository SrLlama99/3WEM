## Excepciones

### try - except
PAra la gestión de excepciones tendremos un bloque "try" y otro "except". 
- try: en el bloque "try" se ubica el código que puede generar una excepción.
- except: en el bloque "except" se ubica el código que va a gestionar la excepción.

El error que suceda en el fragmento de código entre el try y el except no terminará la ejecución del programa. El programa continuará en la primera línea de código después del except y no se ejecutarán más líneas del bloque try. 

El bloque except sólo se ejecuta si hay una excepción en el bloque try. 

Sintaxis:

try:

	# Es un lugar donde
	# se puede generar 
    # una excepción.

except:

	# Es un espacio dedicado  
    # a gestionar la excepción. 

```python
try:
    value = int(input('Ingresa un número natural: '))
    print('El resultado de 1/valor introducido es: ', 1/value)        
except:
    print('No se que hacer con el valor introducido')
```

** En este caso se están generando excepciones de diferentes tipos según valores introducidos. Si introducimos 0 el error es ZeroDivisionError y si el valor de entrada es una cadena de caracteres o un float el error es ValueError. 

### Gestión de varias excepciones

Se pueden indicar varias excepciones indicándoles un nombre de excepción específico. Sólo una rama "except" se ejecuta. Se pueden indicar muchas excepciones pero ninguna de las excepciones se puede especificar más de una vez. 

```python
try:
    value = int(input('Ingresa un número natural: '))
    print('El resultado de 1/valor introducido es: ', 1/value)        
except ValueError:
    print('El valor introducido no es un entero')
except ZeroDivisionError:
    print('El valor introducido es cero')
```

```python
try:
    value = int(input('Ingresa un número natural: '))
    print('El resultado de 1/valor introducido es: ', 1/value)        
except (ValueError, ZeroDivisionError) as e:
    print(f"Ocurrió un error: {e}")
```

### Excepción "por defecto"

Se puede incluir un except por defecto al final para gestionar el resto de excepciones que no tienen un except específico.

```python
try:
    value = int(input('Ingresa un número natural: '))
    print('El resultado de 1/valor introducido es: ', 1/value)        
except ValueError:
    print('El valor introducido no es un entero')
# except ZeroDivisionError:
#    print('El valor introducido es cero')
except:
    print('El valor introducido no es correcto')
```

### Excepciones más comunes

- ZeroDivisionError. cuando intentas realizar una operación que provoque una división en la que el divisor es cero. 
- ValueError. cuando estás manejando valores que pueden usarse de manera inapropiada en algún contexto
- TypeError. intentas aplicar un dato cuyo tipo no se puede aceptar en el contexto actual
- AttributeError. Uso de un método que que no existe en un elemento con el que se está tratando.  
- SyntaxError. Error que se genera por ser un lenguaje interpretado.
- Otros: https://www.w3schools.com/python/python_ref_exceptions.asp

### else

SE puede definir un bloque "else" que se ejecuta si no hay excepción 

```python
try:
  print("Hola")
except:
  print("...error...")
else:
  print("...sin error...")
```

### finally

Se puede definir un bloque "finally" que se ejecuta si hay o si no hay error. USo para liberar recursos o cerrar archivos.

```python
try:
  print("Hola")
except:
  print("...error...")
else:
    print("...sin error...")
finally:
  print("...FIN...")
```

### Generar una excepción
El usuario puede lanzar una excepción utilizando la palabra reservada "raise".

```python
try:
    x = "Hola"

    if not type(x) is int:
      raise Exception("Sólo se permiten valores enteros")
        
except Exception:
    print("Generada excepción de usuario")
```

```python
try:
    x = "Hola"

    if not type(x) is int:
      raise TypeError("Sólo se permiten valores enteros")
        
except TypeError:
    print("Generada excepción de usuario")
```

#### EJEMPLOS

```python
try:
    # Código que puede generar un error
    resultado = 10 / 0
except ZeroDivisionError as e:
    # Captura un código de error específico
    print(f"Ocurrió un error: {e}")
except (ValueError, TypeError) as e:
    # Captura múltiples excepciones
    print(f"Error de valor o tipo: {e}")
except Exception as e:
    # Captura cualquier otra excepción genérica
    print(f"Error inesperado: {e}")
else:
    # Se ejecuta si no ocurre ninguna excepción
    print("Operación exitosa")
finally:
    # Se ejecuta siempre, haya o no error (ideal para limpieza)
    print("Fin del bloque")
```

### Buenas prácticas excepciones

- Evita usar except: sin especificar el tipo de excepción, ya que puede ocultar errores inesperados.
- Maneja solo las excepciones que esperas y puedes tratar de manera significativa.
- Usa finally para liberar recursos importantes (archivos, conexiones de base de datos, etc.).
