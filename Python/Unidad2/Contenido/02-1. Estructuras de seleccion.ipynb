#### if / elif / else

Sintaxis:

    if '<'expresion'>': 

        <sentencia(s)>

'<'expresion'>': Expresión booleana. Su resultado será True o False
<sentencia(s)>: Cuando el resultado de la expresión es True se ejecutan estas sentencia(s). 

Es IMPORTANTE respetar la sangría para indicar el bloque de sentencias a ejecutar cuando se cumple la expresión. 


````python
num = 3
if num < 4:
    print ('El número es menor a 4')
````

Sintaxis:

    if '<'expresion'>': 

        <sentencia(s)>

    else : 

        <sentencia(s)>

Si la expresión es False, se ejecutarán las senticas del bloque "else".

````python
num = 3
if num < 4:
    print ('El número es menor a 4')
else:
    print('El número es mayor o igual a 4')
````

Sintaxis:

    if '<'expresion1'>': 

        <sentencia(s)>

    elif '<'expresion'>': 

        <sentencia(s)>
    ...
    else : 

        <sentencia(s)>

Si la expresión1 es False, se comprobará la expresión del siguiente elif. Si esta es False continuará al siguiente elif. Si no hay más se ejecutarán las sentencias del bloque "else", si es que se han definido.

````python
a = 3
b = 3
if b > a: #Expresión con FAlSE
  print("b es mayor a a")
elif a == b: #exoresión evaluada con TRUE
  print("a es igual a b")
````

````python
a = 3
b = 3
if b > a:  #Expresión con FAlSE
  print("b es mayor a a")
elif a < b: #Expresión con FAlSE
  print("a es mayor a b")
else:
  print("a es igual a b")
````

Se puede indicar la sentencia if en una única línea.

  ```if <expresion>: <sentencia1>; <sentencia2>; <sentenciaN>```

  ```if <expresion>: <sentencia1>; <sentencia2>; <sentenciaN>```

  ```else: <sentencia1>; <sentencia2>```


````python
if num <4: print('El número es menor a 4')
````

````python
num = 5
if num <4: print('El número es menor a 4')
else: print('El número es mayor o igual a 4')
````

#### Match

Python ofrece algo similar al switch de otros lenguajes de programación desde la versión 3.10

Se trata del *match*. Cada *case* define un camino posible. El _ es la opción por defecto, que se ejecuta si la entrada no coincide con ningún caso.



````python
dia = 2

match dia:
    case 1:
        print("Lunes")
    case 2:
        print("Martes")
    case 3:
        print("Miércoles")
    case _:
        print("Día no válido")  # Caso por defecto (comodín)
````

También podemos tener en nuestros case múltiples condiciones, donde *|* es interpretado como un *or*


````python
mes = 4
match mes:
    case 12 | 1 | 2: print("Invierno")
    case 3 | 4 | 5: print("Primavera")
    case 6 | 7 | 8: print("Verano")
    case 9 | 10 | 11: print("Otoño")
    case _: print("Error")
````
